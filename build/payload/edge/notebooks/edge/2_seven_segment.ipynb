# 4-digit 7-segment display

The display is multiplexed in the FPGA logic, so it shows a steady value without any help
from Python. You can show numbers, hex values, short text or your own segment patterns.

```python
from pynq.overlays.base import BaseOverlay
base = BaseOverlay("base.bit")
seg = base.seven_seg
```

```python
seg.show_number(1234)
```

```python
seg.show_number(-3.75)
```

```python
seg.show_hex(0xBEEF)
```

```python
seg.show_text("HELP")
```

Your own patterns: bit 0 = segment a, 1 = b, 2 = c, 3 = d, 4 = e, 5 = f, 6 = g, 7 = decimal point.

```python
seg.show_segments([0x01, 0x40, 0x08, 0x80])
```

## A counter

```python
import time
for n in range(0, 101):
    seg.show_number(n)
    time.sleep(0.05)
```

## A clock (hh.mm) for 20 seconds

```python
end = time.time() + 20
while time.time() < end:
    t = time.localtime()
    seg.show_text("{:02d}.{:02d}".format(t.tm_hour, t.tm_min))
    time.sleep(0.5)
```

```python
seg.clear()
```
