# Buzzer

Fit the jumper on **J10 between pins 1 and 2** to connect the buzzer.

```python
from pynq.overlays.base import BaseOverlay
base = BaseOverlay("base.bit")
bz = base.buzzer
```

```python
bz.beep(2000, 0.3)
```

```python
C4, D4, E4, F4, G4, A4, B4, C5 = 262, 294, 330, 349, 392, 440, 494, 523
bz.play([(C4, .25), (D4, .25), (E4, .25), (F4, .25), (G4, .25), (A4, .25), (B4, .25), (C5, .5)])
```

## Buzz when a push button is pressed (10 seconds)

```python
import time
end = time.time() + 10
while time.time() < end:
    b = base.buttons.read()
    if b == 1:
        bz.tone(1000)
    elif b == 2:
        bz.tone(2000)
    else:
        bz.off()
    time.sleep(0.02)
bz.off()
```
