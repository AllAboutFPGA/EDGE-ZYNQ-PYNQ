# Episode 2 — First boot and a Jupyter tour

## Getting the board online

1. Write `EDGE_ZYNQ_PYNQ_v3.1.1_*.img` to a **16 GB or larger** microSD card with balenaEtcher
   (it can open the .zip directly).
2. Set the board to boot from the microSD card, insert the card and connect an Ethernet cable to your router.
3. Power on. After about a minute the LEDs blink, `EdGE` appears, and then the 7-segment display
   shows the board's IP address, one number at a time, and keeps the last number on.
4. In a browser open `http://<board IP>:9090` and log in with the password **xilinx**.

**No router?** Connect the Ethernet cable straight to your PC and give the PC a fixed address such as
`192.168.2.1`. The board is always reachable at **192.168.2.99** as well.

## Jupyter basics

* **Shift + Enter** runs a cell and moves to the next one.
* A cell is either **code** (Python) or **markdown** (text like this).
* `[*]` next to a cell means it is still running; **Kernel ▸ Interrupt** stops it.
* **Kernel ▸ Restart** gives you a fresh Python (variables are lost, the hardware keeps its state).

## What is this computer?

```python
import os, platform, shutil

print("Linux:", platform.release(), "on", platform.machine())
print("Python:", platform.python_version())
print("CPU cores:", os.cpu_count())

with open("/proc/meminfo") as f:
    total_kb = int(f.readline().split()[1])
print("Memory: {:.0f} MB".format(total_kb / 1024))

disk = shutil.disk_usage("/")
print("SD card: {:.1f} GB used of {:.1f} GB".format(disk.used / 1e9, disk.total / 1e9))

with open("/proc/uptime") as f:
    print("Up for {:.0f} seconds".format(float(f.read().split()[0])))
```

## The terminal

**File ▸ New ▸ Terminal** opens a Linux shell in the browser. Useful commands:

| Command | What it does |
| --- | --- |
| `ls`, `cd`, `cat` | list folders, change folder, show a file |
| `ip addr` | network addresses |
| `top` | running programs and CPU load (press `q` to quit) |
| `sudo poweroff` | shut down safely before you switch off the power (password `xilinx`) |

Always shut down with `sudo poweroff` before removing power: pulling the plug while Linux is
writing can damage the SD card.

## Moving files to and from the board

* **Download:** right-click a file in the file browser ▸ Download.
* **Network share:** on Windows open `\\<board IP>\xilinx` (user `xilinx`, password `xilinx`).
